```python
# Define the function "calculate_days"
from datetime import datetime
import re
def calculate_days(text):
    #get str range into %Y, %M, %D
    pattern = (
    r"(?i)(?:jan|feb|mar|apr|may|jun|jul|aug|sep|oct|nov|dec)\s+\d{1,2},\s+\d{4}"
    )
    # 1. Find all matching date strings
    date_strings = re.findall(pattern, text)
    #print(date_strings)  # Output: ['jun 18, 2022', 'aug 05, 2023']

    # 2. Convert all found strings into datetime objects using a list comprehension
    datetime_objects = [
        datetime.strptime(date, "%b %d, %Y") for date in date_strings
    ]

    # Print the resulting datetime objects
    #for dt in datetime_objects:
     #   print(dt)

    delta = datetime_objects[0] - datetime_objects[1]
    return abs(delta.days)
```

```python
phrase = 'Aired: Jun 18, 2022 - Aug 28, 2022'
# test that your function works

calculate_days(phrase)
```
